```python
# dict

fruit_protein = {
    "Avocado": 2.0,       # grams of protein
    "Guava": 2.6,
    "Blackberries": 2.0,
    "Oranges": 0.9,
    "Banana": 1.1,
    "Apples": 0.3,
    "Kiwi": 1.1,
    "Pomegranate": 1.7,
    "Mango": 0.8,
    "Cherries": 1.0
}
```

```python
s2  = pd.Series(fruit_protein, name = "Protein")
s2
```

```python
s2.loc[['Avocado', 'Apples']]
```

```python
s2.iloc[[1,2,3,4]]
```

# b-Conditional Selection

```python
s2 > 1  
```

```python
s2[s2>1]
```

 # c-Logical Operations   (and - or - not) 

```python
[(s2 > 0.5) & (s2 < 2)]
```

```python
s2[(s2 > 0.5) & (s2 < 2)]
```

```python
s2[(s2>0.5) | (s2 <= 2)]
```

```python
# not operation:

s2[~(s2>1)]
```

# DATAFRAME

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"],
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
}
```

```python
df = pd.DataFrame(data)
df
```
